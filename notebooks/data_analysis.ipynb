```python
import sys
sys.path.append('../src/')
import glamos_processing as glamos
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
landsat = glamos.get_data(1984, 2025)
landsat
```

```python
landsat.id.unique()
```

```python
counts = landsat.id.value_counts().reset_index()
order = landsat.id.value_counts().index.to_numpy()
custom_colors = ['#e63946' if count < 5 else '#457b9d' for count in counts['count']]
```

```python
sns.set_theme(style="whitegrid")
```

```python
f, ax = plt.subplots(figsize=(16,8))
f.tight_layout()
ax.set_title('Observations per Glacier (1984-2025)', fontsize=16, fontweight='bold', pad=15)
ax.set_xlabel('Observation Count', fontsize=12, fontweight='bold', labelpad=15)
ax.set_ylabel('Glacier ID', fontsize=12, fontweight='bold', labelpad=15)
sns.barplot(counts, y='id', x='count', order=order, palette=custom_colors, hue='id',ax=ax)
sns.despine(left=True, bottom=True, ax=ax)
```

```python
f.savefig('landsat_dist.svg', format='svg', bbox_inches='tight', transparent=True)
```
